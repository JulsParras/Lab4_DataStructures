```python
LAST_NAME = "Parras"
SEED_NUM = 3597 #from id TUPM-26-3597
FAVORITE_ARTIST = "Daughtry"

seed_sum = sum(int(digit) for digit in str(SEED_NUM))
artist = len(FAVORITE_ARTIST)

artist_positions = [
    ord(char) for char in FAVORITE_ARTIST
]

T1_data = {}

for i, category in enumerate(categories):
    artist_value = artist_positions[i % len(artist_positions)]

    value = (seed_sum + artist_value + i) % 100

    T1_data[category] = value

print("T1:", test_sources["T1"])

for category, values in T1_data.items():
    print(f"  {category}: {values}")



```
