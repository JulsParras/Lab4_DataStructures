```python
LAST_NAME = "Parras"
SEED_NUM = 3579
FAVORITE_ARTIST = "Daughtry"

# products & prices 
product_prices = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 500,
    "CHARGER": 300}

# customer names
customers = [
    f"{LAST_NAME}_C{i}"
    for i in range(1, 6)]

seed_sum = sum(int(digit) for digit in str(SEED_NUM))
artist_positions = [ord(char) for char in FAVORITE_ARTIST]


transactions = []
for customer_index, customer in enumerate(customers):
    for product_index, (product, price) in enumerate(product_prices.items()):

        #products, customer
        if (customer_index + product_index) % 2 == 0:
            artist_value = artist_positions[
                (customer_index + product_index) % len(artist_positions)]

            quantity = (
                seed_sum + artist_value + customer_index + product_index
            ) % 5 + 1

            transactions.append({
                "customer": customer,
                "product": product,
                "quantity": quantity,
                "price": price,
                "transaction_value": quantity * price})

# Add one repeated customer-product transaction
transactions.append(transactions[0].copy())

# Print Transactions by Customer
for customer in customers:
    print(f"\n{customer}:")

    for transaction in transactions:
        if transaction["customer"] == customer:
            print(
                f"  {transaction['product']}: "
                f"Quantity = {transaction['quantity']}, "
                f"Price = {transaction['price']}, "
                f"Value = {transaction['transaction_value']}"
            )
```

```python
# quantity of buy frm customers
product_quantities = {}

for product in product_prices:
    product_quantities[product] = sum(
        transaction["quantity"]
        for transaction in transactions
        if transaction["product"] == product
    )

print("TOTAL SOLD:")

for product, quantity in product_quantities.items():
    print(f"{product}: {quantity}")
```

```python
# Total money from nb
notebook_total = sum(
    transaction["transaction_value"]
    for transaction in transactions
    if transaction["product"] == "NOTEBOOK")

print("NOTEBOOK TOTAL SALES: ₱", notebook_total)

# Total money from pen
pen_total = sum(
    transaction["transaction_value"]
    for transaction in transactions
    if transaction["product"] == "PEN"
)

print("PEN TOTAL SALES: ₱", pen_total)


# Total from usb
usb_total = sum(
    transaction["transaction_value"]
    for transaction in transactions
    if transaction["product"] == "USB")

print("USB TOTAL SALES: ₱", usb_total)

# Total from headphones
headphones_total = sum(
    transaction["transaction_value"]
    for transaction in transactions
    if transaction["product"] == "HEADPHONES")

print("HEADPHONES TOTAL SALES: ₱", headphones_total)


# Total from charger
charger_total = sum(
    transaction["transaction_value"]
    for transaction in transactions
    if transaction["product"] == "CHARGER")

print("CHARGER TOTAL SALES: ₱", charger_total)
```

```python
# Total 
overall_total = sum(
    transaction["transaction_value"]
    for transaction in transactions
)

print("TOTAL SALES: ₱", overall_total)
```
